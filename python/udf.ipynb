```python
def calculater(a,b):
    addition = a+b
    subtraction = a - b
    return addition,subtraction
x,y = calculater(20,20)
print("addition:",x)
print("subtraction:",y)
```

```python
def find_sum(num):
    total = 0
    for num in num:
        total+=num
    return total
num = [10,20,30,40]
print(find_sum(num))
```

```python
def display_student(student):
    for key,value in student.items():
        print(key,":",value)
student = {
    "name":"rahul",
    "age":21,
    "course":"python"
}
display_student(student)
```
