```python
class student:
    def show(self):
        print("i am a student")
student1 = student()
student2 = student()
student1.show()
student2.show()
```

```python
class student:
    def display(self):
        self.name = "john"
        self.age = 20
        print("name:", self.name)
        print("age:", self.age)
student1 = student()
student1.display()
        
```
