```python
class student:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def show(self):
        print("name:", self.name)
        print("age:", self.age)
student1 = student("john", 20)
student1.show()
```

```python
class employee:
    def __init__(self,name,salary):
        self.name = name
        self.salary = salary
    def display(self):
        print("name:",self.name)
        print("salary:",self.salary)
employee1 = employee("alice", 50000)    
employee1.display()
```

```python
class car:
    def __init__(self, brand, color, price):
        self.brand = brand
        self.color = color
        self.price = price
    def display(self):
        print("brand:",self.brand)
        print("color:",self.color)
        print("price:",self.price)
car1 = car("toyota","red",50000)
car2 = car("kiya","black",75000)
car3 = car("bmw","white",300000)
car1.display()
car2.display()
car3.display()
```

```python
class student:
    def __init__(self,name,roll_no,marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks  
    def display(self):
        print("name:",self.name)
        print("roll_no:",self.roll_no)
        print("marks:",self.marks)
student1 = student("john", 101, 85)
student2 = student("alice", 102, 90)
student1.display()
student2.display()
```

```python
class employee:
    def __init__(self,name,deparment,salary):
        self.name = name
        self.department = deparment
        self.salary = salary
    def display(self):
        print("name:",self.name)
        print("department:",self.department)
        print("salary:",self.salary)
employee1 = employee("john", "IT", 50000)
employee2 = employee("alice", "HR", 60000)  
employee1.display()
employee2.display()
```

```python
class product:
    def __init__(self,name,price):
        self.name = name
        self.price = price
    def display(self):
        print("name:",self.name)
        print("price:",self.price)
product1 = product("laptop", 40000) 
product1.display()


```

```python
class bankaccount:
    def __init__(self,account_holder,balance):
        self.account_holder = account_holder
        self.balance = balance
    def display(self):
        print("account_holder:",self.account_holder)
        print("balance:",self.balance)
bankaccount1 = bankaccount("granthi", 10000000)
bankaccount1.display()  
```
